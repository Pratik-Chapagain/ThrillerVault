```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Create TF-IDF vectors
vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=5000,
    ngram_range=(1, 2)
)

tfidf_matrix = vectorizer.fit_transform(df['plot'])
cosine_sim = cosine_similarity(tfidf_matrix)

print(f"✅ Similarity matrix created! Shape: {cosine_sim.shape}")
```

```python
def get_recommendations(title, df, cosine_sim, top_n=5):
    """Get movie recommendations based on plot similarity"""
    
    title_lower = title.lower()
    matches = df[df['title'].str.lower() == title_lower]
    
    if len(matches) == 0:
        print(f"❌ Movie '{title}' not found!")
        return []
    
    idx = matches.index[0]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:top_n+1]
    
    recommendations = []
    for i, score in sim_scores:
        movie_data = {
            'title': df.iloc[i]['title'],
            'year': df.iloc[i]['release_year'],
            'director': df.iloc[i]['director'],
            'genres': df.iloc[i]['genres'],
            'similarity_score': score
        }
        recommendations.append(movie_data)
    
    return recommendations

print("✅ Recommendation function created!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def create_similarity_heatmap(movie_titles):
    """Create a heatmap showing similarity between selected movies"""
    
    selected_movies = []
    for title in movie_titles:
        matches = df[df['title'].str.lower() == title.lower()]
        if len(matches) > 0:
            selected_movies.append(matches.iloc[0]['title'])
        else:
            print(f"⚠️ Movie '{title}' not found")
    
    if len(selected_movies) < 2:
        print("❌ Need at least 2 movies for a heatmap")
        return
    
    indices = []
    for title in selected_movies:
        idx = df[df['title'] == title].index[0]
        indices.append(idx)
    
    sim_matrix = cosine_sim[indices][:, indices]
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(sim_matrix,
                xticklabels=selected_movies,
                yticklabels=selected_movies,
                cmap='coolwarm',
                annot=True,
                fmt='.2f',
                square=True)
    
    plt.title('Movie Similarity Heatmap')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# Create heatmap for some movies
create_similarity_heatmap(["se7en", "zodiac", "prisoners", "gone girl", "shutter island"])
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Load the embedding model (downloads once, caches for later)
print("🔄 Loading embedding model...")
model = SentenceTransformer('all-MiniLM-L6-v2')

print(f"✅ Model loaded!")
print(f"📊 Embedding dimension: {model.get_sentence_embedding_dimension()}")
```

```python
# Step 3: Create embeddings for all movie plots
print("🔄 Creating embeddings for all movies...")
movie_embeddings = model.encode(df['plot'].tolist(), show_progress_bar=True)

print(f"✅ Created {len(movie_embeddings)} embeddings!")
print(f"📊 Shape: {movie_embeddings.shape}")  # (80, 384)
```

```python
def get_embeddings_recommendations(title, df, embeddings, top_n=5):
    """
    Get movie recommendations using semantic embeddings
    (Understands meaning, not just keywords!)
    """
    
    # Find the movie (case-insensitive)
    title_lower = title.lower()
    matches = df[df['title'].str.lower() == title_lower]
    
    if len(matches) == 0:
        print(f"❌ Movie '{title}' not found!")
        return []
    
    idx = matches.index[0]
    
    # Get the embedding for this movie
    movie_embedding = embeddings[idx].reshape(1, -1)
    
    # Calculate similarity with ALL movies
    similarities = cosine_similarity(movie_embedding, embeddings)[0]
    
    # Get top N similar movies (skip first - it's the movie itself)
    similar_indices = similarities.argsort()[-top_n-1:][::-1][1:top_n+1]
    
    recommendations = []
    for i in similar_indices:
        movie_data = {
            'title': df.iloc[i]['title'],
            'year': df.iloc[i]['release_year'],
            'director': df.iloc[i]['director'],
            'genres': df.iloc[i]['genres'],
            'similarity_score': similarities[i]
        }
        recommendations.append(movie_data)
    
    return recommendations

print("✅ Embedding recommender function created!")
```

```python
def compare_recommendations(title, df, cosine_sim, embeddings, top_n=5):
    """
    Compare TF-IDF and Embedding recommendations side by side
    """
    
    print(f"\n🔍 Comparing recommendations for '{title}'")
    print("=" * 70)
    
    # Get TF-IDF recommendations
    tfidf_recs = get_recommendations(title, df, cosine_sim, top_n)
    
    # Get Embedding recommendations
    embedding_recs = get_embeddings_recommendations(title, df, embeddings, top_n)
    
    print("\n📊 TF-IDF Recommendations (Keyword-based):")
    print("-" * 45)
    if tfidf_recs:
        for i, movie in enumerate(tfidf_recs, 1):
            print(f"{i}. {movie['title']} ({movie['year']}) - {movie['similarity_score']:.3f}")
    else:
        print("No recommendations found")
    
    print("\n🧠 Embeddings Recommendations (Meaning-based):")
    print("-" * 45)
    if embedding_recs:
        for i, movie in enumerate(embedding_recs, 1):
            print(f"{i}. {movie['title']} ({movie['year']}) - {movie['similarity_score']:.3f}")
    else:
        print("No recommendations found")

# Compare for "Se7en"
compare_recommendations("Se7en", df, cosine_sim, movie_embeddings)
```

```python
def natural_language_search(query, df, embeddings, top_n=5):
    """
    Search movies using natural language description
    
    Parameters:
    - query: User's description (e.g., "dark psychological thriller")
    - df: DataFrame with movie data
    - embeddings: Movie embeddings
    - top_n: Number of results to return
    
    Returns:
    - List of matching movies with similarity scores
    """
    
    # Convert user query to embedding
    query_embedding = model.encode([query])
    
    # Calculate similarity with all movies
    similarities = cosine_similarity(query_embedding, embeddings)[0]
    
    # Get top matches
    similar_indices = similarities.argsort()[-top_n:][::-1]
    
    results = []
    for i in similar_indices:
        movie_data = {
            'title': df.iloc[i]['title'],
            'year': df.iloc[i]['release_year'],
            'director': df.iloc[i]['director'],
            'genres': df.iloc[i]['genres'],
            'plot': df.iloc[i]['plot'][:100] + '...',  # Shortened plot
            'similarity_score': similarities[i]
        }
        results.append(movie_data)
    
    return results

print("✅ Natural language search function created!")
```

```python
import matplotlib.pyplot as plt

def visualize_search_results(query, df, embeddings, top_n=5):
    """
    Visualize natural language search results
    """
    
    results = natural_language_search(query, df, embeddings, top_n)
    
    if not results:
        print("No results found!")
        return
    
    # Prepare data
    movie_names = [m['title'] for m in results]
    scores = [m['similarity_score'] for m in results]
    
    # Create bar chart
    plt.figure(figsize=(10, 6))
    bars = plt.bar(movie_names, scores, color='skyblue', edgecolor='navy')
    
    # Add score labels on bars
    for bar, score in zip(bars, scores):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
                f'{score:.3f}', ha='center', va='bottom', fontsize=10)
    
    plt.xlabel('Movies')
    plt.ylabel('Match Score')
    plt.title(f'Search Results: "{query}"')
    plt.xticks(rotation=45, ha='right')
    plt.ylim(0, 1)
    plt.tight_layout()
    plt.show()

# Test visualization
visualize_search_results("psychological thriller about serial killers", df, movie_embeddings)
```

```python
# ============================================
# TMDB API FUNCTIONS
# ============================================

TMDB_BASE_URL = "https://api.themoviedb.org/3"

def search_movie_tmdb(title, year=None):
    """
    Search for a movie on TMDB by title
    """
    url = f"{TMDB_BASE_URL}/search/movie"
    params = {
        "api_key": TMDB_API_KEY,
        "query": title,
        "language": "en-US"
    }
    
    if year:
        params["year"] = year
    
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()
        data = response.json()
        
        if data.get('results'):
            return data['results'][0]
        else:
            return None
            
    except Exception as e:
        print(f"❌ Error searching for '{title}': {e}")
        return None

def get_movie_details_tmdb(movie_id):
    """
    Get detailed information for a movie
    """
    url = f"{TMDB_BASE_URL}/movie/{movie_id}"
    params = {
        "api_key": TMDB_API_KEY,
        "language": "en-US",
        "append_to_response": "credits"
    }
    
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()
        return response.json()
        
    except Exception as e:
        print(f"❌ Error getting details: {e}")
        return None

def get_poster_url(poster_path, size="w500"):
    """
    Get full URL for movie poster
    """
    if poster_path:
        return f"https://image.tmdb.org/t/p/{size}{poster_path}"
    return None

def get_director(movie_details):
    """Extract director from movie details"""
    credits = movie_details.get('credits', {})
    crew = credits.get('crew', [])
    
    for person in crew:
        if person.get('job') == 'Director':
            return person.get('name')
    return None

def get_cast(movie_details, limit=5):
    """Extract cast members from movie details"""
    credits = movie_details.get('credits', {})
    cast = credits.get('cast', [])
    
    names = []
    for actor in cast[:limit]:
        names.append(actor.get('name'))
    
    return ', '.join(names)

def enrich_movie_with_tmdb(title, year=None):
    """
    Get ALL TMDB data for a movie
    """
    # Search for the movie
    movie_result = search_movie_tmdb(title, year)
    
    if not movie_result:
        return None
    
    # Get detailed info
    movie_id = movie_result['id']
    details = get_movie_details_tmdb(movie_id)
    
    if not details:
        return None
    
    # Extract useful fields
    enriched_data = {
        'tmdb_id': details.get('id'),
        'imdb_id': details.get('imdb_id'),
        'title': details.get('title'),
        'release_year': int(details.get('release_date', '').split('-')[0]) if details.get('release_date') else None,
        'runtime': details.get('runtime'),
        'vote_average': details.get('vote_average'),
        'vote_count': details.get('vote_count'),
        'popularity': details.get('popularity'),
        'tagline': details.get('tagline'),
        'overview': details.get('overview'),
        'poster_path': details.get('poster_path'),
        'poster_url': get_poster_url(details.get('poster_path')),
        'backdrop_path': details.get('backdrop_path'),
        'genres': ', '.join([g['name'] for g in details.get('genres', [])]),
        'director': get_director(details),
        'cast': get_cast(details, 5),
    }
    
    return enriched_data

print("✅ All TMDB functions ready!")
```

```python
def enrich_all_movies(df, delay=0.5):
    """
    Enrich all movies with TMDB data
    """
    
    print("🔄 Enriching all movies with TMDB data...")
    print("=" * 60)
    
    # Create new columns
    df['tmdb_id'] = None
    df['imdb_id'] = None
    df['runtime'] = None
    df['vote_average'] = None
    df['vote_count'] = None
    df['popularity'] = None
    df['tagline'] = None
    df['poster_url'] = None
    df['enriched'] = False
    
    success_count = 0
    fail_count = 0
    
    for idx, row in df.iterrows():
        title = row['title']
        year = row.get('release_year')
        
        print(f"🔍 #{idx+1}: {title} ({year})", end=" ")
        
        try:
            data = enrich_movie_with_tmdb(title, year)
            
            if data:
                df.at[idx, 'tmdb_id'] = data.get('tmdb_id')
                df.at[idx, 'imdb_id'] = data.get('imdb_id')
                df.at[idx, 'runtime'] = data.get('runtime')
                df.at[idx, 'vote_average'] = data.get('vote_average')
                df.at[idx, 'vote_count'] = data.get('vote_count')
                df.at[idx, 'popularity'] = data.get('popularity')
                df.at[idx, 'tagline'] = data.get('tagline')
                df.at[idx, 'poster_url'] = data.get('poster_url')
                df.at[idx, 'enriched'] = True
                
                print(f"✅ Found! (Rating: {data.get('vote_average', 'N/A')})")
                success_count += 1
            else:
                print("❌ Not found")
                fail_count += 1
                
        except Exception as e:
            print(f"❌ Error: {e}")
            fail_count += 1
        
        # Be polite to the API
        time.sleep(delay)
    
    print("\n" + "=" * 60)
    print(f"📊 Enrichment Summary:")
    print(f"   ✅ Successful: {success_count}")
    print(f"   ❌ Failed: {fail_count}")
    print(f"   📚 Total: {len(df)} movies")
    
    return df

# Run enrichment
df_enriched = enrich_all_movies(df, delay=0.5)
```

```python
def show_movie_poster(title, df):
    """Display a movie poster"""
    row = df[df['title'].str.lower() == title.lower()]
    if not row.empty:
        poster_url = row.iloc[0].get('poster_url')
        if poster_url:
            print(f"🎬 {title}")
            display(Image(url=poster_url, width=200))
        else:
            print(f"❌ No poster found for '{title}'")
    else:
        print(f"❌ Movie '{title}' not found")

# Show posters for these movies
movies_to_show = ["Se7en", "Jaws", "Psycho", "Inception", "The Dark Knight"]
for movie in movies_to_show:
    show_movie_poster(movie, df_enriched)
```
